```python
import pandas as pd
from pathlib import Path

ROOT_DIR = "./"
TARGET_DIRS = ["DeepSeek", "Gemma", "Kimi", "MetaLlama", "Qwen"]   # 👈 only these folders

OUTPUT_FILE = "dataset.csv"

def merge_selected_dirs(root_dir, target_dirs, output_file):
    root = Path(root_dir)

    all_dfs = []

    for folder in target_dirs:
        folder_path = root / folder

        if not folder_path.exists():
            print(f"⚠️ Skipping missing folder: {folder_path}")
            continue

        for csv_file in folder_path.rglob("*.csv"):
            try:
                df = pd.read_csv(csv_file)
                all_dfs.append(df)
                print(f"Loaded: {csv_file}")
            except Exception as e:
                print(f"Skipped {csv_file}: {e}")

    if not all_dfs:
        print("No CSV files found!")
        return

    
    merged_df = pd.concat(all_dfs, ignore_index=True)

    merged_df.to_csv(output_file, index=False)

    print(f"\n✅ Merged into {output_file}")
    print(f"Total rows: {len(merged_df)}")


# Run
merge_selected_dirs(ROOT_DIR, TARGET_DIRS, OUTPUT_FILE)
```
